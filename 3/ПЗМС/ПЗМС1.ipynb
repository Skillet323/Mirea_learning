```python
import numpy as np 
Real_data = [159, 190, 176, 185, 181, 198, 174, 177, 192, 170, 169, 180, 168, 175, 161, 168, 180, 173, 183, 173, 167, 172, 175, 148, 168, 164, 187, 140, 181]
Integer_data = [5, 7, 6, 5, 12, 6, 7, 10, 7, 12, 6, 4, 9, 4]  

```

```python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

# Данные
Integer_data = [5, 7, 6, 5, 12, 6, 7, 10, 12, 6, 4, 9, 4]
n = len(Integer_data)

print("=== ЦЕЛОЧИСЛЕННЫЕ ДАННЫЕ ===")
```

```python
# Сортировка данных
sorted_data = sorted(Integer_data)
print(f"Отсортированные данные: {sorted_data}")

# Подсчет частот
freq_counter = Counter(Integer_data)
unique_values = sorted(freq_counter.keys())

# Создание таблицы частот
print("\nВариационный ряд:")
print("Значение | Абс. частота | Отн. частота")
print("-" * 40)
abs_freq = []
rel_freq = []
for val in unique_values:
    abs_f = freq_counter[val]
    rel_f = abs_f / n
    abs_freq.append(abs_f)
    rel_freq.append(rel_f)
    print(f"{val:8} | {abs_f:12} | {rel_f:.4f}")
```

```python
# Построение полигона относительных частот
plt.figure(figsize=(10, 6))
plt.plot(unique_values, rel_freq, 'o-', linewidth=2, markersize=8)
plt.xlabel('Значения')
plt.ylabel('Относительная частота')
plt.title('Полигон относительных частот')
plt.grid(True, alpha=0.3)
plt.xticks(unique_values)
plt.show()
```

```python
# Эмпирическая функция распределения
def empirical_cdf(data, x_vals):
    n = len(data)
    sorted_data = sorted(data)
    cdf_values = []
    for x in x_vals:
        count = sum(1 for val in sorted_data if val <= x)
        cdf_values.append(count / n)
    return cdf_values

# Создание точек для графика
x_range = np.linspace(min(Integer_data) - 1, max(Integer_data) + 1, 1000)
y_cdf = empirical_cdf(Integer_data, x_range)

# Построение графика
plt.figure(figsize=(10, 6))
plt.plot(x_range, y_cdf, linewidth=2)
plt.xlabel('Значения')
plt.ylabel('F_n(x)')
plt.title('Эмпирическая функция распределения')
plt.grid(True, alpha=0.3)
plt.show()

# Таблица значений ЭФР
print("\nЭмпирическая функция распределения:")
print("x    | F_n(x)")
print("-" * 15)
cumulative_count = 0
for val in sorted(freq_counter.keys()):
    cumulative_count += freq_counter[val]
    cdf_val = cumulative_count / n
    print(f"{val:4} | {cdf_val:.4f}")
```

```python
# Вычисление статистик
data_array = np.array(Integer_data)

mean_val = np.mean(data_array)
variance_val = np.var(data_array)
std_val = np.std(data_array)
median_val = np.median(data_array)
cv_val = (std_val / mean_val) * 100

print("\n=== ОПИСАТЕЛЬНЫЕ СТАТИСТИКИ ===")
print(f"Выборочное среднее: {mean_val:.4f}")
print(f"Выборочная дисперсия: {variance_val:.4f}")
print(f"Выборочное стандартное отклонение: {std_val:.4f}")
print(f"Выборочная медиана: {median_val:.4f}")
print(f"Коэффициент вариации: {cv_val:.2f}%")
```

```python
# Вещественные данные
Real_data = [159, 190, 176, 185, 181, 198, 174, 177, 192, 170, 169, 180, 168, 175, 161, 168, 180, 173, 183, 173, 167, 172, 175, 148, 168, 164, 187, 140, 181]
n_real = len(Real_data)

print("\n\n=== ВЕЩЕСТВЕННЫЕ ДАННЫЕ ===")
```

```python
# Правило Стёрджесса
k = 1 + 3.322 * np.log10(n_real)
k_rounded = round(k)
print(f"Число групп по правилу Стёрджесса: {k:.2f} ≈ {k_rounded}")
```

```python
# Вычисление границ групп
min_val = min(Real_data)
max_val = max(Real_data)
range_val = max_val - min_val
bin_width = range_val / k_rounded

print(f"\nМинимум: {min_val}")
print(f"Максимум: {max_val}")
print(f"Размах: {range_val}")
print(f"Ширина интервала: {bin_width:.2f}")

# Создание границ интервалов
bins = np.linspace(min_val, max_val, k_rounded + 1)
print(f"\nГраницы интервалов:")
for i in range(len(bins) - 1):
    print(f"Интервал {i+1}: [{bins[i]:.2f}, {bins[i+1]:.2f})")
```

```python
# Подсчет частот в интервалах
hist, bin_edges = np.histogram(Real_data, bins=bins)

print("\nВариационный ряд (интервальные данные):")
print("Интервал           | Середина | Абс. частота | Отн. частота")
print("-" * 60)
rel_freq_hist = []
midpoints = []
for i in range(len(hist)):
    left = bin_edges[i]
    right = bin_edges[i+1]
    midpoint = (left + right) / 2
    midpoints.append(midpoint)
    abs_freq = hist[i]
    rel_freq = abs_freq / n_real
    rel_freq_hist.append(rel_freq)
    print(f"[{left:6.2f}, {right:6.2f}) | {midpoint:8.2f} | {abs_freq:12} | {rel_freq:.4f}")
```

```python
# Построение гистограммы относительных частот
plt.figure(figsize=(10, 6))
plt.hist(Real_data, bins=bins, weights=np.ones(n_real) / n_real, 
         edgecolor='black', alpha=0.7)
plt.xlabel('Значения')
plt.ylabel('Относительная частота')
plt.title('Гистограмма относительных частот')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Эмпирическая функция распределения для вещественных данных
x_range_real = np.linspace(min(Real_data) - 5, max(Real_data) + 5, 1000)
y_cdf_real = empirical_cdf(Real_data, x_range_real)

plt.figure(figsize=(10, 6))
plt.plot(x_range_real, y_cdf_real, linewidth=2)
plt.xlabel('Значения')
plt.ylabel('F_n(x)')
plt.title('Эмпирическая функция распределения (вещественные данные)')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Вычисление статистик для вещественных данных
real_array = np.array(Real_data)

mean_real = np.mean(real_array)
variance_real = np.var(real_array)
std_real = np.std(real_array)
median_real = np.median(real_array)
cv_real = (std_real / mean_real) * 100

print("\n=== ОПИСАТЕЛЬНЫЕ СТАТИСТИКИ (ВЕЩЕСТВЕННЫЕ ДАННЫЕ) ===")
print(f"Выборочное среднее: {mean_real:.4f}")
print(f"Выборочная дисперсия: {variance_real:.4f}")
print(f"Выборочное стандартное отклонение: {std_real:.4f}")
print(f"Выборочная медиана: {median_real:.4f}")
print(f"Коэффициент вариации: {cv_real:.2f}%")
```
