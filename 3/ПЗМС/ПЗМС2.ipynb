```python
import numpy as np
import pandas as pd


data1 = np.loadtxt("data/1.txt")
data2 = np.loadtxt("data/2.txt")
data3 = np.loadtxt("data/3.txt")
data4 = np.loadtxt("data/4.txt")
```

```python
import math as m
print("РЯД 1")
print(data1)
N = len(data1)
mean = sum(data1)/N
std_sq = m.sqrt(sum((x - mean)**2 for x in data1) / (N - 1))
X_095 = 1.96
tr_int_1 = [
    float(mean - X_095 * (std_sq / m.sqrt(N))),
    float(mean + X_095 * (std_sq / m.sqrt(N)))
]
print("N-1 = ",N-1 )
t_99_0005 = 2.872
tr_int_2 = [
    float(mean - t_99_0005 * (std_sq / m.sqrt(N))),
    float(mean + t_99_0005 * (std_sq / m.sqrt(N)))
]
chi2_lower = 74.22
chi2_upper = 129.5  

lower_bound_var = (m.sqrt(N-1) * std_sq) / m.sqrt(chi2_upper)
upper_bound_var = (m.sqrt(N-1) * std_sq) / m.sqrt(chi2_lower)

tr_int_std = [
    lower_bound_var,
    upper_bound_var
]

print("\nВЫЧИСЛЕНИЯ ДЛЯ РЯДА 1:")
print("Среднее отклонение:", mean)
print("Среднеквадратичное отклонение:", std_sq)
print("Доверительный интервал с известным отклонением:", tr_int_1)
print("Доверительный интервал с неизвестным отклонением:", tr_int_2)
print("Доверительный интервал χ²-распределение:", tr_int_std )
```
