```python
import csv
from datetime import datetime
import matplotlib.pyplot as plt
import pandas as pd

data = pd.read_csv('sensor_data.csv')

data.head()
```

```python
# === 1. Линейный график — температура ===
plt.figure(figsize=(12, 4))
plt.plot(data['timestamp'], data['temperature'], color='red', marker='o', markersize=2)
plt.title("Температура во времени")
plt.xlabel("Время")
plt.ylabel("°C")
plt.grid(True)

# Отображаем метки по оси X каждую 10-ю запись
step = 10
tick_indices = range(0, len(data), step)
plt.xticks(tick_indices, [data['timestamp'].iloc[i] for i in tick_indices], rotation=45)

plt.tight_layout()
```

```python
# === 2. Гистограмма — напряжение ===

plt.hist(data['voltage'], bins=10, color='steelblue', edgecolor='black')
plt.title("Распределение напряжения")
plt.xlabel("Вольты")
plt.ylabel("Частота")
plt.grid(axis='y')
```

```python
# === 3. Круговая диаграмма — движением ===
plt.figure(figsize=(8, 8))
counts = data["motion"].value_counts()
plt.pie(counts, labels=counts.index, autopct='%1.1f%%')
plt.title("Доля времени с движением")

plt.tight_layout()
plt.savefig("pr8_visualization.png")
plt.show()
```
