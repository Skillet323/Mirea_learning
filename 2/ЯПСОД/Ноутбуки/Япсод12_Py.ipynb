```python
# Листинг 1 – EDA и корреляция в Python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("mxmh_survey_results.csv")
df.dropna(inplace=True)

# Описательная статистика
print(df[["Age", "Anxiety", "Depression", "Hours per day"]].describe())

# Корреляционная матрица
corr_matrix = df[["Age", "Anxiety", "Depression", "Hours per day"]].corr()
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
plt.title("Корреляционная матрица")
plt.show()
```

```python
# Листинг 3 – Линейная регрессия в Python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

X = df[["Depression", "Hours per day", "Age"]]
y = df["Anxiety"]

model = LinearRegression().fit(X, y)
print(f"R²: {model.score(X, y):.2f}")
print(f"Коэффициенты: {model.coef_}")
```

```python
# Листинг 5 – Логистическая регрессия в Python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

X = pd.get_dummies(df[["Age", "Anxiety", "Depression"]])
y = df["Music effects"]

model = LogisticRegression(max_iter=1000).fit(X, y)
print(classification_report(y, model.predict(X)))
```

```python
# Листинг 7 – K-means кластеризация
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

X = df[["Anxiety", "Depression", "Hours per day"]]
kmeans = KMeans(n_clusters=3).fit(X)
df["Cluster"] = kmeans.labels_

# Визуализация
plt.scatter(df["Anxiety"], df["Depression"], c=df["Cluster"], cmap="viridis")
plt.title("Кластеры пользователей")
plt.show()
```
