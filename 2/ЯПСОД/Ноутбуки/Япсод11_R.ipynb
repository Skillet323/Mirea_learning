```r
library(shiny)
library(ggplot2)

# Загрузка данных
df <- read.csv("train_time_series.csv")

# Проверка, что данные загружены корректно
if (!"family" %in% colnames(df)) {
  stop("В данных отсутствует столбец 'family'. Проверьте файл train_time_series.csv.")
}
if (!"date" %in% colnames(df)) {
  stop("В данных отсутствует столбец 'date'. Проверьте файл train_time_series.csv.")
}
if (!"sales" %in% colnames(df)) {
  stop("В данных отсутствует столбец 'sales'. Проверьте файл train_time_series.csv.")
}

# Преобразование столбца date в формат даты
df$date <- as.Date(df$date)

# UI
ui <- fluidPage(
  titlePanel("Анализ продаж"),
  sidebarLayout(
    sidebarPanel(
      selectInput("family", "Выберите категорию:", 
                  choices = unique(df$family))
    ),
    mainPanel(
      plotOutput("salesPlot") # Используем plotOutput для ggplot2
    )
  )
)

# Server
server <- function(input, output) {
  output$salesPlot <- renderPlot({
    # Фильтрация данных по выбранной категории
    df_filtered <- subset(df, family == input$family)
    
    # Построение графика
    ggplot(df_filtered, aes(x = date, y = sales)) + 
      geom_line(color = "blue") + 
      labs(title = paste("Продажи:", input$family),
           x = "Дата",
           y = "Продажи") +
      theme_minimal()
  })
}

# Запуск приложения
shinyApp(ui, server)
```
