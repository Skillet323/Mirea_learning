```r
library(readr)
df <- read_csv("combined_data.csv")
head(df)
```

```r
#Подсчет пропусков
colSums(is.na(df))
```

```r
#Удалим пропуски
df$Condition[is.na(df$Condition)] = "Unknown"
df = filter(df,!is.na(Mileage))
df = filter(df,!is.na(`Monthly Payment`))
```

```r
#Удаление дубликатов
df = df[!duplicated(df), ]
print("Дубликаты удалены.")
```

```r
#Преобразование типа
df$`Price (USD)` = as.numeric(gsub("[\\$,]", "", df$`Price (USD)`))
```
