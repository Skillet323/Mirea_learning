```python
#Загрузка набора данных
import pandas as pd
df = pd.read_csv("combined_data.csv")
df.head()
```

```python
#Подсчет пустых строк
print(df.isnull().sum())
```

```python
#Преобразование строковых данных в численные
import numpy as np
df['Mileage'] = df['Mileage'].apply(
    lambda x: float(x[0:-4].replace(",", ".")) if isinstance(x, str) and x.endswith(" mi.") else np.nan
)
df['Monthly Payment'] = df['Monthly Payment'].apply(
    lambda x: float(x[6:-3].replace(",", "")) if isinstance(x, str) and x.endswith("/mo") else np.nan
)

```

```python
#Заполнение численных пропусков средним значением и медианой. А строковые новой строкой
df['Mileage'] = df['Mileage'].fillna(df['Mileage'].mean())
df['Monthly Payment'] = df['Monthly Payment'].fillna(df['Monthly Payment'].median())
df['Condition'] = df['Condition'].fillna("Unknown")
```

```python
#Удаление дубликатов
df = df.drop_duplicates()
print("Дубликаты удалены.")
```

```python
#Конечное преобразование строковых данных
df['Price (USD)'] = df['Price (USD)'].apply(
    lambda x: float(x[1::].replace(",", "")) if isinstance(x, str) and x.startswith("$") else np.nan
)
df
```
