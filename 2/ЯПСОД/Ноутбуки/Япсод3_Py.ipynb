```python
#Загрузка набора данных
import pandas as pd
df = pd.read_csv("mxmh_survey_results.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Гистограмма
plt.figure(figsize=(8, 6))
sns.histplot(df['Age'], bins=20, kde=True)
plt.title("Гистограмма возрастов участников опроса")
plt.show()
```

```python
df["Date"] = df["Timestamp"].apply(lambda x: ".".join(x.split("/")[:2]))
plt.figure(figsize=(8, 6))
#Линейный график
plt.plot(df['Date'].unique()[:15], df.groupby('Date')['Insomnia'].mean()[:15], marker='o')
plt.title("Линейный график среднего значения бессоницы по датам")
plt.xlabel("Дата")
plt.ylabel("Бессоница")
plt.show()
```

```python
#Коробка с усами
plt.figure(figsize=(12, 5))
sns.boxplot(x='Fav genre', y='Hours per day', data=df)
plt.title("Боксплот часов прослушивания в день по любимым жанрам")
plt.show()
```
