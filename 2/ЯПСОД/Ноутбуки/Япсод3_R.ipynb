```r
library(readr)
df <- read_csv("mxmh_survey_results.csv")
head(df)
```

```r
library(ggplot2)
ggplot(df, aes(x = Age)) +
geom_histogram(bins = 20, fill = "blue", color = "black", alpha
= 0.7) +
labs(title = "Гистограмма возрастов")
```

```r
ggplot(head(df,n=12), aes(x = Timestamp, y = Insomnia, group=1)) +
geom_line(color="blue") +
geom_point() +
labs(title = "Линейный график бессоницы по датам", x = "Дата", y
= "Бессоница")
```

```r
ggplot(df, aes(x = `Fav genre`, y = `Hours per day`)) +
geom_boxplot(fill = "orange", color = "black") +
labs(title = "Боксплот часов прослушивания в день по любимым жанрам")
```
