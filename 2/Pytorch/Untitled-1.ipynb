```python
import torch
a = torch.tensor([2.0], requires_grad=True)
b = torch.tensor([4.0], requires_grad=True)
c = torch.tensor([1.0], requires_grad=True)
d = torch.tensor([5.0], requires_grad=False)

print(a.grad)
print(b.grad)
print(c.grad)
print(d.grad)

def show_tensor_params(*tensors):
  for x in tensors:
    print('---')
    print(f"data - {x.data}")
    print(f"grad - {x.grad}")
    print(f"grad_fn - {x.grad_fn}")
    print(f"req_grad - {x.requires_grad}")
    print(f"is_leaf - {x.is_leaf}")

s = (a,b,c,d)
for x in s:
  print (show_tensor_params(x))
```

```python
loss = a*b + c*d
loss.backward()
for x in s:
    print(x.grad)
print (loss)
```
