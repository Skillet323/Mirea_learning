```python
def main(number):
    # Список для хранения результатов
    result = []
    
    # Проверяем каждый бит от 0 до 23
    for bit in range(24):
        # Создаем маску для проверки текущего бита
        mask = 1 << bit
        
        # Если текущий бит установлен (равен 1)
        if number & mask:
            # Добавляем имя и значение битового поля в результат
            result.append((f'T{bit}', str(bit)))
    
    return result

# Примеры тестов
print(main(1875564))  # [(T0, '0'), (T1, '1'), (T2, '3'), (T4, '114')]
print(main(6400607))  # [(T0, '3'), (T1, '3'), (T2, '3'), (T4, '390')]
print(main(9179457))  # [(T0, '1'), (T1, '0'), (T2, '0'), (T4, '560')]
print(main(13219396)) # [(T0, '0'), (T1, '1'), (T2, '1'), (T4, '806')]
```

```python
# Блок 2.1. Реализация методов чтения, записи и размера
class MyDict:
    def __init__(self):
        self.items = []

    def __getitem__(self, key):
        for k, v in self.items:
            if k == key:
                return v
        raise KeyError(f"Key {key} not found")

    def __setitem__(self, key, value):
        for i, (k, _) in enumerate(self.items):
            if k == key:
                self.items[i] = (key, value)
                return
        self.items.append((key, value))

    def __len__(self):
        return len(self.items)

# Тесты для 2.1
d = MyDict()
d['a'] = 1
d['b'] = 2
print(d['a']) # Чтение
print(len(d)) # Размер
try:
    d['c']  # Несуществующий ключ
except KeyError:
    print("Нет такого ключа")
    pass
else:
    print( False, "KeyError не сгенерирован")




```

```python
# Блок 2.2. Интерфейс, как в dict (del, in, etc)
class MyDict(MyDict):  # Наследуем предыдущий класс
    def __delitem__(self, key):
        for i, (k, _) in enumerate(self.items):
            if k == key:
                del self.items[i]
                return
        raise KeyError(f"Key {key} not found")

    def __contains__(self, key):
        for k, _ in self.items:
            if k == key:
                return True
        return False

# Тесты для 2.2
def test_2_2():
    d = MyDict()
    d['a'] = 1
    assert 'a' in d  # Проверка in
    del d['a']       # Удаление
    assert 'a' not in d
    try:
        del d['a']   # Удаление несуществующего ключа
    except KeyError:
        pass
    else:
        assert False, "KeyError не сгенерирован"

test_2_2()
print("2.2: Тесты пройдены")




```

```python
# Блок 2.3. Итератор для цикла for
class MyDict(MyDict):  # Наследуем предыдущий класс
    def __iter__(self):
        seen = set()
        for k, _ in self.items:
            if k not in seen:
                seen.add(k)
                yield k
# Тесты для 2.3
def test_2_3():
    d = MyDict()
    d['a'] = 1
    d['b'] = 2
    d['a'] = 3  # Проверка дублирующихся ключей

    # Проверка простой итерации
    keys = [k for k in d]
    assert keys == ['a', 'b'], "Итератор не перечисляет все ключи"

    # Проверка вложенных циклов
    for key in d:
        for _ in range(2):
            assert key in d, "Итерация внутри цикла нарушена"

test_2_3()
print("2.3: Тесты пройдены")
```

```python
# Блок 3.1. Классы узлов дерева выражений
class Num:
    def __init__(self, value):
        self.value = value

class Add:
    def __init__(self, left, right):
        self.left = left
        self.right = right

class Mul:
    def __init__(self, left, right):
        self.left = left
        self.right = right

# Тест для 3.1
def test_3_1():
    ast = Add(Num(7), Mul(Num(3), Num(2)))
    assert isinstance(ast, Add)
    assert isinstance(ast.left, Num)
    assert isinstance(ast.right, Mul)
    assert ast.left.value == 7
    assert ast.right.left.value == 3
    print("3.1: Тест пройден")

test_3_1()




```

```python
# Блок 3.2. Visitor для печати выражений
class PrintVisitor:
    def visit(self, node):
        method_name = f'visit_{type(node).__name__}'
        method = getattr(self, method_name, None)
        if not method:
            raise TypeError(f'No visit method for {type(node).__name__}')
        return method(node)

    def visit_Num(self, node):
        return str(node.value)

    def visit_Add(self, node):
        return f"({self.visit(node.left)} + {self.visit(node.right)})"

    def visit_Mul(self, node):
        return f"({self.visit(node.left)} * {self.visit(node.right)})"

# Тест для 3.2
def test_3_2():
    ast = Add(Num(7), Mul(Num(3), Num(2)))
    pv = PrintVisitor()
    assert pv.visit(ast) == '(7 + (3 * 2))'
    print("3.2: Тест пройден")

test_3_2()
```

```python
# Блок 3.3. Visitor для вычисления выражений
class CalcVisitor:
    def visit(self, node):
        method_name = f'visit_{type(node).__name__}'
        method = getattr(self, method_name, None)
        if not method:
            raise TypeError(f'No visit method for {type(node).__name__}')
        return method(node)

    def visit_Num(self, node):
        return node.value

    def visit_Add(self, node):
        return self.visit(node.left) + self.visit(node.right)

    def visit_Mul(self, node):
        return self.visit(node.left) * self.visit(node.right)

# Тест для 3.3
def test_3_3():
    ast = Add(Num(7), Mul(Num(3), Num(2)))
    cv = CalcVisitor()
    assert cv.visit(ast) == 13
    print("3.3: Тест пройден")

test_3_3()
```

```python
# Блок 3.4. Visitor для генерации кода стековой машины
class StackVisitor:
    def __init__(self):
        self.code = []

    def visit(self, node):
        method_name = f'visit_{type(node).__name__}'
        method = getattr(self, method_name, None)
        if not method:
            raise TypeError(f'No visit method for {type(node).__name__}')
        method(node)
        return '\n'.join(self.code)

    def visit_Num(self, node):
        self.code.append(f"PUSH {node.value}")

    def visit_Add(self, node):
        self.visit(node.left)
        self.visit(node.right)
        self.code.append("ADD")

    def visit_Mul(self, node):
        self.visit(node.left)
        self.visit(node.right)
        self.code.append("MUL")

# Тест для 3.4
def test_3_4():
    ast = Add(Num(7), Mul(Num(3), Num(2)))
    sv = StackVisitor()
    expected_code = '''\
PUSH 7
PUSH 3
PUSH 2
MUL
ADD'''
    assert sv.visit(ast).strip() == expected_code.strip()
    print("3.4: Тест пройден")

ast = Add(Mul(Num(2), Add(Num(3), Num(4))),Mul(Num(3),Add(Num(1),Num(2))))
sv = StackVisitor( )
print(sv.visit(ast).strip())
```

```python
# Блок 4.1. HTML-теги через менеджер контекста
class HTML:
    def __init__(self):
        self.code = []
        self.indent = 0

    def __getattr__(self, tag_name):
        # Возвращаем функцию-обертку, которая создает менеджер контекста
        def tag_context_manager():
            return self.TagContextManager(self, tag_name)
        return tag_context_manager

    class TagContextManager:
        def __init__(self, html_obj, tag):
            self.html = html_obj
            self.tag = tag

        def __enter__(self):
            self.html.code.append(f"{'  ' * self.html.indent}<{self.tag}>")
            self.html.indent += 1

        def __exit__(self, exc_type, exc_val, exc_tb):
            self.html.indent -= 1
            self.html.code.append(f"{'  ' * self.html.indent}</{self.tag}>")

    def p(self, text):
        self.code.append(f"{'  ' * self.indent}<p>{text}</p>")

    def get_code(self):
        return '\n'.join(self.code)

# Тест
html = HTML()
with html.body():
    with html.div():
        with html.div():
            html.p('Первая строка.')
            html.p('Вторая строка.')
        with html.div():
            html.p('Третья строка.')

print("Генерируемый HTML:")
print(html.get_code())
```

```python
# Блок 5.1. Библиотека для рисования SVG
class SVG:
    def __init__(self):
        self.elements = []
        self.width = 0
        self.height = 0

    def line(self, x1, y1, x2, y2, color='black'):
        self.elements.append(f'<line x1="{x1:.6f}" y1="{y1:.6f}" x2="{x2:.6f}" y2="{y2:.6f}" stroke="{color}" />')

    def circle(self, cx, cy, r, color='red'):
        self.elements.append(f'<circle cx="{cx:.6f}" cy="{cy:.6f}" r="{r:.6f}" fill="{color}" />')

    def save(self, filename, width, height):
        self.width = width
        self.height = height
        with open(filename, 'w') as f:
            f.write(f'<svg version="1.1" width="{width:.6f}" height="{height:.6f}" xmlns="http://www.w3.org/2000/svg">\n')
            for elem in self.elements:
                f.write(f'  {elem}\n')
            f.write('</svg>')

# Пример использования
svg = SVG()
svg.line(10, 10, 60, 10, color='black')
svg.line(60, 10, 60, 60, color='black')
svg.line(60, 60, 10, 60, color='black')
svg.line(10, 60, 10, 10, color='black')
svg.circle(10, 10, r=5, color='red')
svg.circle(60, 10, r=5, color='red')
svg.circle(60, 60, r=5, color='red')
svg.circle(10, 60, r=5, color='red')
svg.save('pic.svg', 100, 100)
print("SVG-файл сохранен как 'pic.svg'")
```

```python
# Блок 5.2. Алгоритм Д. Кнута для бинарных деревьев
scale_x = 25
scale_y = 50

class Tree:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
        self.x = 0
        self.y = 0

class Position:
    def __init__(self):
        self.current_x = 0

def assign_coordinates(tree, y_level, pos):
    if tree is None:
        return
    assign_coordinates(tree.left, y_level + 1, pos)
    tree.x = pos.current_x * scale_x
    tree.y = y_level * scale_y
    pos.current_x += 1
    assign_coordinates(tree.right, y_level + 1, pos)

def draw_tree_svg(root, filename="pic_5_2.svg"):
    svg = SVG()
    all_nodes = []
    stack = [(root, False)]
    while stack:
        node, visited = stack.pop()
        if not node:
            continue
        if visited:
            all_nodes.append(node)
            if node.left:
                svg.line(node.x, node.y, node.left.x, node.left.y)
            if node.right:
                svg.line(node.x, node.y, node.right.x, node.right.y)
            svg.circle(node.x, node.y, r=5, color='blue')
        else:
            stack.append((node, True))
            stack.append((node.right, False))
            stack.append((node.left, False))
    
    min_x = min(n.x for n in all_nodes) - 10
    max_x = max(n.x for n in all_nodes) + 10
    max_y = max(n.y for n in all_nodes) + 10
    width = max_x - min_x + 20
    height = (max_y // scale_y + 2) * scale_y
    
    svg.save(filename, width, height)
    print(f"5.2: SVG-файл '{filename}' сохранен. Координаты корня: ({root.x}, {root.y})")

# Пример использования
tree_2 = Tree(2, Tree(3, Tree(4), Tree(5)), Tree(6, Tree(7)))
tree_8 = Tree(8, Tree(9, Tree(10), Tree(11, Tree(12), Tree(13))), Tree(14))
tree = Tree(1, tree_2, tree_8)
pos = Position()
assign_coordinates(tree, 0, pos)
draw_tree_svg(tree)
print(f"5.2: Координаты корня: x={tree.x}, y={tree.y}")
print(f"Координаты левого потомка: x={tree.left.x}, y={tree.left.y}")
print(f"Координаты правого потомка: x={tree.right.x}, y={tree.right.y}\n")
```

```python
# Блок 5.3. Улучшенный алгоритм для бинарных деревьев
def layout_binary_tree(root, y_level=0, x_offset=0):
    if not root:
        return (0, 0)
    left_x, left_right = layout_binary_tree(root.left, y_level + 1, x_offset)
    right_x, right_right = layout_binary_tree(root.right, y_level + 1, left_right)
    
    if root.left and root.right:
        root.x = (left_right + right_x) / 2
        current_right = right_right
    elif root.left:
        root.x = left_right + scale_x
        current_right = root.x + scale_x
    elif root.right:
        root.x = right_x - scale_x
        current_right = right_right
    else:
        root.x = x_offset
        current_right = root.x + scale_x
    
    root.y = y_level * scale_y
    return (root.x, current_right)

def draw_tree_svg(root, filename="pic_5_3.svg"):
    svg = SVG()
    all_nodes = []
    stack = [(root, False)]
    while stack:
        node, visited = stack.pop()
        if not node:
            continue
        if visited:
            all_nodes.append(node)
            if node.left:
                svg.line(node.x, node.y, node.left.x, node.left.y)
            if node.right:
                svg.line(node.x, node.y, node.right.x, node.right.y)
            svg.circle(node.x, node.y, r=5, color='green')
        else:
            stack.append((node, True))
            stack.append((node.right, False))
            stack.append((node.left, False))
    
    max_right = layout_binary_tree(root)[1]
    min_x = min(n.x for n in all_nodes) - 10
    max_x = max_right + 10
    max_y = max(n.y for n in all_nodes) + 10
    width = max_x - min_x + 20
    height = (max_y // scale_y + 2) * scale_y
    
    svg.save(filename, width, height)
    print(f"5.3: SVG-файл '{filename}' сохранен. Координаты корня: ({root.x}, {root.y})")

# Пример использования
tree_2 = Tree(2, Tree(3, Tree(4), Tree(5)), Tree(6, Tree(7)))
tree_8 = Tree(8, Tree(9, Tree(10), Tree(11, Tree(12), Tree(13))), Tree(14))
tree = Tree(1, tree_2, tree_8)
layout_binary_tree(tree)
draw_tree_svg(tree)
print(f"5.3: Улучшенные координаты корня: x={tree.x}, y={tree.y}")
print(f"Координаты левого потомка: x={tree.left.x}, y={tree.left.y}")
print(f"Координаты правого потомка: x={tree.right.x}, y={tree.right.y}\n")
```

```python
# Блок 5.4. Алгоритм для n-арных деревьев
class NTreeNode:
    def __init__(self, val, children=None):
        self.val = val
        self.children = children or []
        self.x = 0
        self.y = 0

def layout_nary_tree(root, y_level=0, x_offset=0, spacing=50):
    if not root:
        return (0, 0)
    children_coords = []
    total_width = 0
    for child in root.children:
        cx, cw = layout_nary_tree(child, y_level + 1, x_offset + total_width, spacing)
        children_coords.append((cx, cw))
        total_width += cw  # Не используется, но остается для совместимости

    if root.children:
        first_child_x = children_coords[0][0]
        last_child_right = children_coords[-1][0] + children_coords[-1][1]
        root.x = (first_child_x + last_child_right) / 2
        total_width = last_child_right - first_child_x
    else:
        root.x = x_offset
        total_width = 1  # Минимальная ширина для листа

    # Исправленный расчет Y
    root.y = y_level * spacing * 2
    return (root.x, total_width)

def draw_nary_tree_svg(root, filename="pic_5_4.svg", spacing=50):
    svg = SVG()
    all_nodes = []
    stack = [(root, False)]
    while stack:
        node, visited = stack.pop()
        if not node:
            continue
        if visited:
            all_nodes.append(node)
            for child in node.children:
                svg.line(node.x, node.y, child.x, child.y)
            svg.circle(node.x, node.y, r=5, color='purple')
        else:
            stack.append((node, True))
            for child in reversed(node.children):
                stack.append((child, False))
    
    # Определение границ
    min_x = min(n.x for n in all_nodes) - 10
    max_x = max(n.x + (n.right_boundary if hasattr(n, 'right_boundary') else 0) for n in all_nodes) + 10
    max_y = max(n.y for n in all_nodes) + 20
    width = max_x - min_x + 20
    height = max_y + 20
    
    svg.save(filename, width, height)
    print(f"5.4: SVG-файл '{filename}' сохранен. Координаты корня: ({root.x}, {root.y})")

# Пример использования
root = NTreeNode("A", [
    NTreeNode("B", [NTreeNode("D"), NTreeNode("E")]),
    NTreeNode("C", [NTreeNode("F"), NTreeNode("G")])
])
layout_nary_tree(root)
draw_nary_tree_svg(root)
print(f"5.4: Координаты n-арного корня: x={root.x}, y={root.y}")
print(f"Координаты первого потомка: x={root.children[0].x}, y={root.children[0].y}\n")
```

```python
# Блок 5.5. Визуализация AST (пример)
import ast

class ASTNode:
    def __init__(self, node):
        self.type = type(node).__name__
        self.children = []
        self.x = 0
        self.y = 0
        for field in ast.iter_fields(node):
            if isinstance(field[1], list):
                for item in field[1]:
                    if isinstance(item, ast.AST):
                        child = ASTNode(item)
                        self.children.append(child)
            elif isinstance(field[1], ast.AST):
                child = ASTNode(field[1])
                self.children.append(child)

def visualize_ast(code, filename="pic_5_5.svg"):
    tree = ast.parse(code)
    root = ASTNode(tree)
    # Применяем простой алгоритм для координат (например, 5.2)
    assign_coordinates_ast(root, 0, Position())
    draw_ast_svg(root, "pic_5_5.svg")
    print(f"5.5: AST-дерево сохранено в '{filename}'. Корневой узел: {root.type}")

def assign_coordinates_ast(node, y_level, pos):
    if not node:
        return
    assign_coordinates_ast(node.children[0] if node.children else None, y_level + 1, pos)
    node.x = pos.current_x * scale_x
    node.y = y_level * scale_y
    pos.current_x += 1
    assign_coordinates_ast(node.children[1] if len(node.children) > 1 else None, y_level + 1, pos)

def draw_ast_svg(root, filename="pic_5_5.svg"):
    def collect_ast_nodes(node):
        lst = []
        stack = [node]
        while stack:
            current = stack.pop()
            if current:
                lst.append(current)
                stack.extend(current.children)
        return lst

    svg = SVG()
    stack = [(root, False)]
    while stack:
        node, visited = stack.pop()
        if not node:
            continue
        if visited:
            for child in node.children:
                svg.line(node.x, node.y, child.x, child.y)
            svg.circle(node.x, node.y, r=5, color='orange')
        else:
            stack.append((node, True))
            for child in reversed(node.children):
                stack.append((child, False))

    nodes = collect_ast_nodes(root)
    min_x = min(n.x for n in nodes)
    max_x = max(n.x for n in nodes)
    max_y = max(n.y for n in nodes)
    width = max_x - min_x + 50
    height = max_y + 50
    svg.save(filename, width, height)
    print(f"5.5: SVG-файл '{filename}' сохранен.")
```

```python
# Блок 5.6. Физическое моделирование графа
import math
from random import randint
from tkinter import Tk, Canvas, Button

CANVAS_WIDTH, CANVAS_HEIGHT = 800, 600
NODE_R = 15
C1, C2, C3, C4 = 2, 50, 20000, 0.1
DELAY = 10

class Vec:
    def __init__(self, x, y):
        self.x = x
        self.y = y

class Node:
    def __init__(self, text):
        self.text = text
        self.targets = []
        self.vec = Vec(0, 0)

    def to(self, *nodes):
        for n in nodes:
            self.targets.append(n)
            n.targets.append(self)
        return self

class Graph:
    def __init__(self):
        self.nodes = []

    def add(self, text):
        self.nodes.append(Node(text))
        return self.nodes[-1]

class GUI:
    def __init__(self, root):
        self.canvas = Canvas(root, width=CANVAS_WIDTH, height=CANVAS_HEIGHT, bg="white")
        self.draw_button = Button(root, text="Draw", command=self.start_draw)
        self.save_button = Button(root, text="Save SVG", command=self.save_svg)
        self.canvas.pack()
        self.draw_button.pack()
        self.save_button.pack()
        self.nodes = None
        self.busy = None

    def draw_node(self, x, y, text):
        self.canvas.create_oval(x-NODE_R, y-NODE_R, x+NODE_R, y+NODE_R, fill="MistyRose2")
        self.canvas.create_text(x, y, text=text)

    def draw_graph(self):
        for n in self.nodes:
            for t in n.targets:
                self.canvas.create_line(n.vec.x, n.vec.y, t.vec.x, t.vec.y)
        for n in self.nodes:
            self.draw_node(n.vec.x, n.vec.y, n.text)

    def start_draw(self):
        self.canvas.delete("all")
        if self.busy:
            root.after_cancel(self.busy)
        random_layout(self.nodes)
        self.animate()

    def save_svg(self):
        svg = SVG()
        for n in self.nodes:
            svg.circle(n.vec.x, n.vec.y, r=NODE_R, color='blue')
            for t in n.targets:
                svg.line(n.vec.x, n.vec.y, t.vec.x, t.vec.y)
        svg.save("pic_5_6.svg", CANVAS_WIDTH, CANVAS_HEIGHT)
        print("5.6: SVG-файл сохранен как 'pic_5_6.svg'")

    def animate(self):
        self.canvas.delete("all")
        for _ in range(DELAY):
            force_layout(self.nodes)
        self.draw_graph()
        self.busy = root.after(5, self.animate)

def random_layout(nodes):
    for n in nodes:
        n.vec.x = randint(NODE_R * 4, CANVAS_WIDTH - NODE_R * 4)
        n.vec.y = randint(NODE_R * 4, CANVAS_HEIGHT - NODE_R * 4)

def f_spring(u, v):
    dx = v.vec.x - u.vec.x
    dy = v.vec.y - u.vec.y
    dist = math.hypot(dx, dy)
    if dist == 0:
        return Vec(0, 0)
    f = C1 * (dist - C2) / dist
    return Vec(f * dx, f * dy)

def f_ball(u, v):
    dx = v.vec.x - u.vec.x
    dy = v.vec.y - u.vec.y
    dist = math.hypot(dx, dy)
    if dist == 0:
        return Vec(0, 0)
    f = C3 / (dist ** 2)
    return Vec(-f * dx, -f * dy)

def force_layout(nodes):
    forces = {n: Vec(0, 0) for n in nodes}
    for u in nodes:
        for v in nodes:
            if u != v:
                if v in u.targets or u in v.targets:
                    f = f_spring(u, v)
                else:
                    f = f_ball(u, v)
                forces[u].x += f.x
                forces[u].y += f.y
    for n in nodes:
        n.vec.x += forces[n].x * C4
        n.vec.y += forces[n].y * C4

# Пример графа
g = Graph()
n1 = g.add("1")
n2 = g.add("2")
n3 = g.add("3")
n4 = g.add("4")
n5 = g.add("5")
n6 = g.add("6")
n7 = g.add("7")
n1.to(n2, n3, n4, n5)
n2.to(n5)
n3.to(n2, n4)
n6.to(n4, n1, n7)
n7.to(n5, n1)

root = Tk()
w = GUI(root)
w.nodes = g.nodes
root.mainloop()
print("5.6: Граф отображен в окне Tkinter.\n")
```

```python
# Операции языка Фортик
OP_NAMES = {0: 'push', 1: 'op', 2: 'call', 3: 'is', 4: 'to', 5: 'exit'}
LIB = {
    '+': lambda vm: vm.stack.append(vm.stack.pop() + vm.stack.pop()),
    '-': lambda vm: vm.stack.append(vm.stack.pop(-2) - vm.stack.pop()),
    '*': lambda vm: vm.stack.append(vm.stack.pop() * vm.stack.pop()),
    '/': lambda vm: vm.stack.append(vm.stack.pop(-2) // vm.stack.pop()),
    '%': lambda vm: vm.stack.append(vm.stack.pop(-2) % vm.stack.pop()),
    '&': lambda vm: vm.stack.append(vm.stack.pop() & vm.stack.pop()),
    '|': lambda vm: vm.stack.append(vm.stack.pop() | vm.stack.pop()),
    '^': lambda vm: vm.stack.append(vm.stack.pop() ^ vm.stack.pop()),
    '<': lambda vm: vm.stack.append(1 if vm.stack.pop(-2) < vm.stack.pop() else 0),
    '>': lambda vm: vm.stack.append(1 if vm.stack.pop(-2) > vm.stack.pop() else 0),
    '=': lambda vm: vm.stack.append(1 if vm.stack.pop() == vm.stack.pop() else 0),
    '<<': lambda vm: vm.stack.append(vm.stack.pop(-2) << vm.stack.pop()),
    '>>': lambda vm: vm.stack.append(vm.stack.pop(-2) >> vm.stack.pop()),
    'if': lambda vm: vm.if_op(),
    'for': lambda vm: vm.for_op(),
    '.': lambda vm: print(vm.stack.pop(), end=''),
    'emit': lambda vm: print(chr(vm.stack.pop()), end=''),
    '?': lambda vm: vm.stack.append(int(input("Input: "))),
    'array': lambda vm: vm.array_op(),
    '@': lambda vm: vm.at_op(),
    '!': lambda vm: vm.store_op(),
}

class VM:
    def __init__(self, code):
        self.stack = []
        self.code = code
        self.pc = 0
        self.scope = {}
        self.scope_stack = [self.scope]
        self.entry = code[0]
        self.call_stack = []
        self.arrays = {}
        self.array_id = 0

    # Реализация операций array, @, !
    def array_op(self):
        if len(self.stack) < 1:
            raise IndexError("Stack underflow: 'array' operation requires 1 value on the stack")
        size = self.stack.pop()
        self.arrays[self.array_id] = [0] * size
        self.stack.append(self.array_id)
        self.array_id += 1

    def at_op(self):
        if len(self.stack) < 2:
            raise IndexError("Stack underflow: '@' operation requires 2 values on the stack")
        index = self.stack.pop()
        arr_id = self.stack.pop()
        self.stack.append(self.arrays[arr_id][index])

    def store_op(self):
        if len(self.stack) < 3:
            raise IndexError("Stack underflow: '!' operation requires 3 values on the stack")
        value = self.stack.pop()
        index = self.stack.pop()
        arr_id = self.stack.pop()
        self.arrays[arr_id][index] = value

    # Реализация операции if
    def if_op(self):
        if len(self.stack) < 3:
            raise IndexError("Stack underflow: 'if' operation requires 3 values on the stack")
        false_addr = self.stack.pop()
        true_addr = self.stack.pop()
        value = self.stack.pop()
        if value != 0:
            self.pc = true_addr
        else:
            self.pc = false_addr

    # Реализация операции for
    def for_op(self):
        if len(self.stack) < 2:
            raise IndexError("Stack underflow: 'for' operation requires 2 values on the stack")
        func_addr = self.stack.pop()
        count = self.stack.pop()
        if count > 0:
            # Push the loop counter back onto the stack for the function
            self.stack.append(count)
            # Save return_pc as pc + 1
            self.call_stack.append(('for', self.pc + 1, count - 1, func_addr))
            self.pc = func_addr
        else:
            self.pc += 1  # Move to the next instruction
    # Шаг выполнения команды
    def step(self):
        cmd = self.code[self.pc]
        op_code = cmd & 7
        arg = cmd >> 3
        op_name = OP_NAMES.get(op_code, 'unknown')

        if op_name == 'push':
            self.stack.append(arg)
        elif op_name == 'op':
            key = list(LIB.keys())[arg]
            func = LIB[key]
            # Ensure the stack has enough elements for the operation
            required_stack_size = 2 if key in ['+', '-', '*', '/', '%', '&', '|', '^', '<', '>', '=', '<<', '>>'] else 1
            if len(self.stack) < required_stack_size:
                raise IndexError(f"Stack underflow: operation '{key}' requires {required_stack_size} values on the stack")
            func(self)
        elif op_name == 'exit':
            if self.call_stack:
                entry = self.call_stack.pop()
                if isinstance(entry, tuple) and entry[0] == 'for':
                    _, return_pc, remaining_count, func_addr = entry
                    if remaining_count > 0:
                        # Повторный вызов функции
                        self.call_stack.append(('for', return_pc, remaining_count - 1, func_addr))
                        self.pc = func_addr
                    else:
                        # Завершение цикла, переход к return_pc
                        self.pc = return_pc
                else:
                    # Обычный возврат из функции
                    self.pc = entry
                    self.scope_stack.pop()
            else:
                # Завершение программы
                self.pc = len(self.code)
            return
        elif op_name == 'call':
            name = str(arg)
            entry = self.scope.get(name)
            if not entry or entry[0] != 'function':
                raise KeyError(f"Function {name} not found or not a function")
            self.call_stack.append(self.pc + 1)
            self.scope_stack.append({})
            self.pc = entry[1]
            return
        elif op_name == 'is':
            name = str(arg)
            self.scope[name] = ('function', self.pc + 1)
        elif op_name == 'to':
            var_name = str(arg)
            value = self.stack.pop()
            self.scope_stack[-1][var_name] = value
        elif op_name == 'array':
            self.array_op()
        elif op_name == '@':
            self.at_op()
        elif op_name == '!':
            self.store_op()
        elif op_name == 'if':
            self.if_op()
        elif op_name == 'for':
            self.for_op()

        # Увеличиваем pc только если команда не была exit
        if op_name != 'exit':
            self.pc += 1

    # Запуск виртуальной машины
    def run(self):
        self.pc = self.entry
        while self.pc < len(self.code):
            self.step()
# Пример для массива
vm = VM([
    0,          # entry point
    16,         # push 2 (размер массива)
    145,        # op 'array' (18*8+1=145)
    0,          # push 0 (ID массива)
    1,          # push 1 (значение)
    0,          # push 0 (индекс)
    161,        # op '!' (20*8+1=161)
    0,          # push 0 (ID массива)
    0,          # push 0 (индекс)
    153,        # op '@' (19*8+1=153)
    121,        # op '.' (15*8+1=121)
    5           # exit
])
vm.run()  # Выведет '0'
# Пример для цикла for
vm = VM([
    80,         # 0: push 10 (итераций)
    56,         # 1: push 7 (адрес функции, будет после is 1)
    113,        # 2: op 'for'
    5,          # 3: exit
    3, 1,       # 4: is '1'
    336,        # 6: push 42
    129,        # 7: op 'emit'
    5           # 8: exit
])
vm.entry = 0
vm.run()

```

```python
# 6.7 Синтаксический разбор
def parse(source):
    tokens = source.strip().split()
    stack = [[]]
    for token in tokens:
        if token == '[':
            new_list = []
            stack[-1].append(new_list)
            stack.append(new_list)
        elif token == ']':
            stack.pop()
        else:
            stack[-1].append(token)
    ast = stack[0][0]

    def build(node):
        if isinstance(node, list):
            if not node:
                return []
            if node[0] in ['is', 'to', 'call', 'op', 'push']:
                return (node[0], *build(node[1:]))
            else:
                return [build(child) for child in node]
        else:
            return node
    return build(ast)
```

```python
#7.1 Генератор документации
import importlib.util
import sys
import os
import inspect
from typing import Optional

def load_module(file_path: str):
    """Загружает модуль из файла."""
    if not os.path.exists(file_path):
        raise FileNotFoundError(f"Файл {file_path} не найден")
        
    if not file_path.endswith('.py'):
        raise ValueError("Файл должен иметь расширение .py")

    module_dir = os.path.dirname(os.path.abspath(file_path))
    module_name = os.path.splitext(os.path.basename(file_path))[0]
    
    # Сохраняем оригинальные состояния
    original_path = sys.path.copy()
    original_cwd = os.getcwd()
    
    try:
        # Меняем рабочую директорию и sys.path
        os.chdir(module_dir)
        sys.path.insert(0, module_dir)
        
        spec = importlib.util.spec_from_file_location(module_name, file_path)
        
        if spec is None:
            raise ImportError(f"Не удалось создать спецификацию для {file_path}")
            
        if spec.loader is None:
            raise ImportError(f"Не найден загрузчик для {file_path}")

        module = importlib.util.module_from_spec(spec)
        sys.modules[module_name] = module
        spec.loader.exec_module(module)
        return module
    except Exception as e:
        raise ImportError(f"Ошибка при загрузке модуля: {str(e)}") from e
    finally:
        # Восстанавливаем оригинальные состояния
        sys.path = original_path
        os.chdir(original_cwd)

def generate_markdown(module) -> str:
    """Генерирует Markdown-документацию для модуля."""
    md = []
    
    # Заголовок модуля и его описание
    md.append(f"# Модуль {module.__name__}\n")
    if module.__doc__:
        md.append(module.__doc__.strip() + '\n')
    
    # Обработка классов
    classes = []
    for name, obj in vars(module).items():
        if inspect.isclass(obj) and obj.__module__ == module.__name__:
            classes.append(obj)
    
    for cls in classes:
        md.append(f"## Класс {cls.__name__}\n")
        if cls.__doc__:
            md.append(cls.__doc__.strip() + '\n')
        
        # Сбор методов класса
        methods = []
        for method_name, method in vars(cls).items():
            if inspect.isfunction(method):
                methods.append(method)
            elif isinstance(method, (staticmethod, classmethod)):
                # Извлекаем функцию из декораторов
                actual_func = method.__func__
                methods.append(actual_func)
        
        # Добавление методов в документацию
        for method in methods:
            sig = inspect.signature(method)
            md.append(f"* **Метод** `{method.__name__}{sig}`\n")
            if method.__doc__:
                md.append(method.__doc__.strip() + '\n')
    
    # Обработка функций модуля
    functions = []
    for name, obj in vars(module).items():
        if inspect.isfunction(obj) and obj.__module__ == module.__name__:
            functions.append(obj)
    
    for func in functions:
        md.append(f"## Функция {func.__name__}\n")
        sig = inspect.signature(func)
        md.append(f"Сигнатура: `{func.__name__}{sig}`\n")
        if func.__doc__:
            md.append(func.__doc__.strip() + '\n')
    
    return '\n'.join(md)

if __name__ == "__main__":
    if len(sys.argv) != 2:
        print("Usage: python generate_docs.py <module_file>")
        sys.exit(1)
    
    file_path = r"C:\Users\borga\Desktop\Stud\IT_One\BPMN_NEW\server\main_server.py"
    module = load_module(file_path)

    markdown = generate_markdown(module)
    print(markdown)
```

```python
#7.2 Визуализация
import os
import sys
from pathlib import Path

def build_hierarchy(root_path: str, prefix: str = "", is_root: bool = True) -> str:
    """Рекурсивно строит текстовую иерархию модулей и пакетов."""
    root = Path(root_path)
    output = []
    
    if not root.exists():
        return ""
    
    # Определяем тип элемента
    if root.is_dir():
        # Пропускаем пакеты Python
        if (root / "__init__.py").exists():
            output.append(f"{prefix}- Пакет: {root.name}/")
        else:
            output.append(f"{prefix}- Каталог: {root.name}/")
        
        # Добавляем отступ для вложенных элементов
        new_prefix = prefix + "  " if is_root else prefix + "│  "
        
        # Обрабатываем содержимое каталога
        children = sorted(root.iterdir())
        for i, item in enumerate(children):
            is_last = (i == len(children) - 1)
            item_prefix = new_prefix + ("└──" if is_last else "├──")
            
            # Рекурсивная обработка элемента
            item_hierarchy = build_hierarchy(
                str(item),
                prefix=item_prefix,
                is_root=False
            )
            
            if item_hierarchy:
                output.append(item_hierarchy)
    else:
        # Обработка файлов
        if root.stem != "__init__":
            output.append(f"{prefix}- Модуль: {root.name}")
    
    return '\n'.join(output)

if __name__ == "__main__":
    if len(sys.argv) != 2:
        print("Usage: python visualize_hierarchy.py <project_root>")
        sys.exit(1)
    
    project_root = r"C:\Users\borga\Desktop\Stud\IT_One\BPMN_NEW\server"
    hierarchy = build_hierarchy(project_root)
    print(f"Иерархия проекта ({project_root}):\n")
    print(hierarchy)
```

```python
from pathlib import Path
import tkinter as tk
import logging

# Настройка логирования
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s [%(levelname)s] %(message)s',
    handlers=[logging.StreamHandler()]
)

SCALE_X = 8  # Увеличен масштаб для лучшей видимости
SCALE_Y = 8

COLORS = [
    (0, 0, 0), (0, 0, 168), (0, 168, 0), (0, 168, 168),
    (168, 0, 0), (168, 0, 168), (168, 84, 0), (168, 168, 168),
    (84, 84, 84), (84, 84, 252), (84, 252, 84), (84, 252, 252),
    (252, 84, 84), (252, 84, 252), (252, 252, 84), (252, 252, 252)
]

def log_command(pos, cmd, details):
    """Логирование команд парсера."""
    logging.debug(f"[PARSE] Pos: {pos:04X} | Cmd: {cmd:02X} | {details}")

def parse_pic(data):
    """Парсер PIC файла согласно спецификации AGI."""
    commands = []
    pos = 5  # Пропускаем заголовок (5 байт)
    current_color = 15  # Белый цвет по умолчанию
    x, y = 0, 0

    logging.info(f"[PARSE] Начинаем парсинг с позиции {pos}, всего {len(data)} байт")

    while pos < len(data):
        if pos >= len(data):
            break

        cmd = data[pos]
        log_command(pos, cmd, f"Обработка команды: 0x{cmd:02X}")

        try:
            if cmd == 0xFF:  # Конец данных [[7]]
                logging.debug("Конец данных")
                break

            elif cmd == 0x02:  # Move (Перемещение)
                if pos + 2 >= len(data):
                    logging.error("Неполная команда перемещения")
                    break
                x = data[pos + 1]
                y = data[pos + 2]
                pos += 3
                logging.debug(f"Перемещение в ({x}, {y})")

            elif cmd == 0x04:  # Draw (Рисование линии)
                if pos + 2 >= len(data):
                    logging.error("Неполная команда рисования")
                    break
                x2 = data[pos + 1]
                y2 = data[pos + 2]
                commands.append(('line', x, y, x2, y2, current_color))
                x, y = x2, y2
                pos += 3
                logging.debug(f"Линия к ({x2}, {y2})")

            elif cmd == 0x06:  # Set Color (Установка цвета)
                if pos + 1 >= len(data):
                    logging.error("Неполная команда цвета")
                    break
                current_color = data[pos + 1] % 16
                pos += 2
                logging.debug(f"Установлен цвет {current_color}")

            elif cmd == 0x08:  # Fill (Заливка)
                if pos + 2 >= len(data):
                    logging.error("Неполная команда заливки")
                    break
                x_fill = data[pos + 1]
                y_fill = data[pos + 2]
                commands.append(('fill', x_fill, y_fill, current_color))
                pos += 3
                logging.debug(f"Заливка в ({x_fill}, {y_fill})")

            elif cmd == 0x0F:  # Draw Relative (Относительное рисование)
                if pos + 2 >= len(data):
                    logging.error("Неполная относительная линия")
                    break
                dx = data[pos + 1]
                dy = data[pos + 2]
                x2 = x + dx
                y2 = y + dy
                commands.append(('line', x, y, x2, y2, current_color))
                x, y = x2, y2
                pos += 3
                logging.debug(f"Относительная линия к ({x2}, {y2})")

            elif cmd == 0xF4:  # Draw Y corner [[7]]
                if pos + 2 >= len(data):
                    logging.error("Неполная команда Y-corner")
                    break
                x = data[pos + 1]
                y = data[pos + 2]
                pos += 3
                logging.debug(f"Y-corner в ({x}, {y})")
                # Реализация рисования углов

            elif cmd == 0xF5:  # Draw X corner [[7]]
                if pos + 2 >= len(data):
                    logging.error("Неполная команда X-corner")
                    break
                x = data[pos + 1]
                y = data[pos + 2]
                pos += 3
                logging.debug(f"X-corner в ({x}, {y})")
                # Реализация рисования углов

            elif cmd == 0xF6:  # Absolute line [[7]]
                if pos + 2 >= len(data):
                    logging.error("Неполная команда абсолютной линии")
                    break
                x2 = data[pos + 1]
                y2 = data[pos + 2]
                commands.append(('line', x, y, x2, y2, current_color))
                x, y = x2, y2
                pos += 3
                logging.debug(f"Абсолютная линия к ({x2}, {y2})")

            else:
                logging.warning(f"Неизвестная команда 0x{cmd:02X} на позиции {pos}")
                pos += 1

        except Exception as e:
            logging.error(f"Ошибка на позиции {pos}: {e}")
            pos += 1

    logging.info(f"Обработано {len(commands)} команд")
    return commands

def draw(pic_path, canvas):
    """Основная функция отрисовки."""
    if not pic_path.exists():
        logging.error("Файл не найден")
        return

    try:
        data = pic_path.read_bytes()
        logging.info(f"Размер файла: {len(data)} байт")

        # Проверка заголовка
        if len(data) < 5:
            logging.error("Файл слишком короткий")
            return

        width = data[0] + (data[1] << 8)
        height = data[2] + (data[3] << 8)
        start_pos = data[4]
        logging.info(f"Заголовок: ширина={width}, высота={height}, стартовая позиция={start_pos}")

        commands = parse_pic(data)

        for cmd in commands:
            action = cmd[0]
            if action == 'line':
                _, x1, y1, x2, y2, color = cmd
                color_hex = '#%02x%02x%02x' % COLORS[color]
                canvas.create_line(
                    x1 * SCALE_X, y1 * SCALE_Y,
                    x2 * SCALE_X, y2 * SCALE_Y,
                    fill=color_hex, width=1
                )
            elif action == 'fill':
                _, x, y, color = cmd
                color_hex = '#%02x%02x%02x' % COLORS[color]
                canvas.create_rectangle(
                    x * SCALE_X, y * SCALE_Y,
                    (x + 1) * SCALE_X, (y + 1) * SCALE_Y,
                    fill=color_hex, outline=color_hex
                )
            canvas.update()

    except Exception as e:
        logging.error(f"Ошибка при чтении или отрисовке файла: {e}")

if __name__ == "__main__":
    root = tk.Tk()
    canvas = tk.Canvas(root, width=160 * SCALE_X, height=170 * SCALE_Y, bg='black')
    canvas.pack()

    pic_path = Path('data/PIC.2')
    draw(pic_path, canvas)
    root.mainloop()
```
