```python
# 5.1
def ham_dist_naive(a: int, b: int) -> int:
    xor = a ^ b
    dist = 0
    while xor:
        dist += xor & 1
        xor >>= 1
    return dist


# Однострочник
def ham_dist(a, b): return bin(a ^ b).count('1')


# Тесты
print(ham_dist_naive(0b10, 0b11))
print(ham_dist_naive(0b1100, 0b0011))

print(ham_dist(0b10, 0b11))
print(ham_dist(0b1100, 0b0011))
```

```python
# 5.2
def encode_val(n_bits: int, repeat: int, value: int) -> int:
    result = 0
    for i in range(n_bits):
        bit = (value >> i) & 1
        result |= (bit * ((1 << repeat) - 1)) << (i * repeat)
    return result


def decode_val(n_bits: int, repeat: int, value: int) -> int:
    result = 0
    for i in range(n_bits):
        chunk = (value >> (i * repeat)) & ((1 << repeat) - 1)
        ones = bin(chunk).count('1')
        result |= (1 if ones > repeat//2 else 0) << i
    return result


message = [815608, 2064837, 2093080, 2063879, 196608, 2067983, 10457031,
           1830912, 2067455, 2093116, 1044928, 2064407, 6262776, 2027968,
           4423680, 2068231, 2068474, 1999352, 1019903, 2093113, 2068439,
           2064455, 1831360, 1936903, 2067967, 2068456]

decoded = [decode_val(8, 3, val) for val in message]
print(decoded)
text = ''.join(chr(c) for c in decoded)
print(text)
```

```python
#5.3
from functools import cache

@cache
def lev_dist(s1: str, s2: str) -> int:
    if not s1:
        return len(s2)
    if not s2:
        return len(s1)
    
    if s1[0] == s2[0]:
        return lev_dist(s1[1:], s2[1:])
    
    return 1 + min(
        lev_dist(s1[1:], s2),     
        lev_dist(s1, s2[1:]),     
        lev_dist(s1[1:], s2[1:])   
    )


# Тест
print(lev_dist('столб', 'слон'))  # 3
```

```python
# 5.4
@cache
def lev_dist_ops(s1: str, s2: str) -> list:
    if not s1:
        return ['вставка'] * len(s2)
    if not s2:
        return ['удаление'] * len(s1)

    if s1[0] == s2[0]:
        return lev_dist_ops(s1[1:], s2[1:])

    del_cost = lev_dist_ops(s1[1:], s2)
    ins_cost = lev_dist_ops(s1, s2[1:])
    sub_cost = lev_dist_ops(s1[1:], s2[1:])

    min_cost = min(len(del_cost), len(ins_cost), len(sub_cost))
    if min_cost == len(del_cost):
        return ['удаление'] + del_cost
    elif min_cost == len(ins_cost):
        return ['вставка'] + ins_cost
    return ['замена'] + sub_cost


# Тест
print(lev_dist_ops('столб', 'слон'))
```

```python
# 5.5
def levenshtein_matrix(x, y):
    m, n = len(x), len(y)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    # Заполняем первую строку и первый столбец
    for i in range(m + 1):
        dp[i][0] = i  # Удаление
    for j in range(n + 1):
        dp[0][j] = j  # Вставка
    # Заполняем остальную матрицу
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            cost = 0 if x[i - 1] == y[j - 1] else 1
            dp[i][j] = min(dp[i - 1][j] + 1,      # Удаление
                           dp[i][j - 1] + 1,      # Вставка
                           dp[i - 1][j - 1] + cost)  # Замена или ничего
    return dp


def lev_dist_gen(x, y):
    dp = levenshtein_matrix(x, y)
    m, n = len(x), len(y)
    i, j = m, n
    operations = []
    # Восстанавливаем операции, двигаясь от конца к началу
    while i > 0 or j > 0:
        if i > 0 and j > 0 and dp[i][j] == dp[i - 1][j - 1] + (0 if x[i - 1] == y[j - 1] else 1):
            if x[i - 1] != y[j - 1]:
                operations.append(f"x[{i - 1}] = y[{j - 1}]")
            i -= 1
            j -= 1
        elif i > 0 and dp[i][j] == dp[i - 1][j] + 1:
            operations.append(f"del x[{i - 1}]")
            i -= 1
        elif j > 0 and dp[i][j] == dp[i][j - 1] + 1:
            operations.append(f"x.insert({i}, y[{j - 1}])")
            j -= 1
    return operations


# Тестирование на примере из задания
print("Тест из условия:")
print('\n'.join(lev_dist_gen('достаток', 'доставка')))
```

```python
def test_lev_dist_gen():
    import random
    import string

    for _ in range(10):
        s1 = list(''.join(random.choices(
            string.ascii_lowercase, k=random.randint(5, 10))))
        s2 = list(''.join(random.choices(
            string.ascii_lowercase, k=random.randint(5, 10))))
        ops = lev_dist_gen(s1, s2)
        # Применяем операции к s1
        modified = list(s1)
        for op in ops:
            try:
                exec(op, {'x': modified, 'y': s2})
            except Exception as e:
                print(f"Error applying {op}: {e}")
                raise
        assert modified == s2, f"Test failed for {s1} -> {s2}. Result: {modified}"
    print("Все тесты пройдены успешно!")


test_lev_dist_gen()
```

```python
# 5.6
from collections import defaultdict


def create_dictionary(file_path):
    dictionary = {}
    with open(file_path, 'r', encoding='utf-8') as file:
        for line in file:
            parts = line.strip().split()
            if len(parts) == 2:
                word, frequency = parts
                dictionary[word] = int(frequency)
    return dictionary


file_path = "words.txt"
dictionary = create_dictionary(file_path)

# Функция вычисления расстояния Левенштейна


def levenshtein_distance(word1, word2):
    m, n = len(word1), len(word2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            cost = 0 if word1[i - 1] == word2[j - 1] else 1
            dp[i][j] = min(dp[i - 1][j] + 1, dp[i][j - 1] +
                           1, dp[i - 1][j - 1] + cost)
    return dp[m][n]

# Функция поиска кандидатов на исправление


def find_candidates(word, dictionary, max_distance):
    candidates = []
    for dict_word in dictionary:
        dist = levenshtein_distance(word, dict_word)
        if dist <= max_distance:
            candidates.append((dict_word, dictionary[dict_word], dist))
    return candidates

# Функция проверки орфографии


def spell(sentence, dictionary):
    words = sentence.split()
    corrected_words = []
    for word in words:
        if word in dictionary:
            corrected_words.append(word)
        else:
            candidates = find_candidates(word, dictionary, 2)
            if candidates:
                candidates.sort(key=lambda x: (x[2], -x[1]))
                corrected_words.append(candidates[0][0])
            else:
                corrected_words.append(word)
    return ' '.join(corrected_words)


# Тестирование
print("Тест из условия:")
print(spell('помоему я напесал усё правильна', dictionary))
```

```python
# 5.7
# Словарь соответствий английских и русских букв
eng_to_rus = {
    'a': 'а', 'b': 'б', 'c': 'ц', 'd': 'д', 'e': 'е', 'h': 'н', 'i': 'и', 'k': 'к',
    'm': 'м', 'n': 'н', 'o': 'о', 'p': 'р', 'r': 'р', 's': 'с', 't': 'т', 'u': 'у',
    'x': 'х', 'y': 'у'
}

# Модифицированное расстояние Левенштейна


def modified_lev_dist(word1, word2):
    m, n = len(word1), len(word2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            cost = 0 if word1[i - 1] == word2[j - 1] else 1
            # Учет замены английской буквы на русскую
            if word1[i - 1] in eng_to_rus and eng_to_rus[word1[i - 1]] == word2[j - 1]:
                cost = 0.5  # Меньшая стоимость
            dp[i][j] = min(dp[i - 1][j] + 1, dp[i][j - 1] +
                           1, dp[i - 1][j - 1] + cost)
            # Учет перестановки соседних символов
            if i > 1 and j > 1 and word1[i - 1] == word2[j - 2] and word1[i - 2] == word2[j - 1]:
                # Стоимость перестановки
                dp[i][j] = min(dp[i][j], dp[i - 2][j - 2] + 1)
    return dp[m][n]

# Функция поиска кандидатов с модифицированным расстоянием


def find_candidates_modified(word, dictionary, max_distance):
    candidates = []
    for dict_word in dictionary:
        dist = modified_lev_dist(word, dict_word)
        if dist <= max_distance:
            candidates.append((dict_word, dictionary[dict_word], dist))
    return candidates

# Улучшенная функция проверки орфографии


def spell_modified(sentence, dictionary):
    words = sentence.split()
    corrected_words = []
    for word in words:
        if word in dictionary:
            corrected_words.append(word)
        else:
            candidates = find_candidates_modified(word, dictionary, 2)
            if candidates:
                # Сортируем по расстоянию, затем по частоте
                candidates.sort(key=lambda x: (x[2], -x[1]))
                corrected_words.append(candidates[0][0])
            else:
                corrected_words.append(word)
    return ' '.join(corrected_words)


# Тестирование
print("Тест с английскими буквами и перестановками:")
print(spell_modified('ëж еж йож ёж ëж', dictionary))
```

```python
import re
from collections import defaultdict


def split_sentence(sentence):
    # Разделяем по запятым
    parts = [p.strip() for p in sentence.split(',') if p.strip()]

    # Если частей меньше 5, делим их равномерно
    if len(parts) < 5:
        # Считаем общее количество слов
        words = sentence.split()
        total_words = len(words)

        # Если слов меньше 5, дополняем пустыми частями
        if total_words < 5:
            parts = words + [''] * (5 - total_words)
        else:
            # Распределяем слова по 5 частям
            chunk_size = max(1, total_words // 5)
            parts = [' '.join(words[i:i + chunk_size])
                     for i in range(0, total_words, chunk_size)]

            # Если получилось больше 5 частей, объединяем последние
            if len(parts) > 5:
                parts[4] = ' '.join(parts[4:])
                parts = parts[:5]

    # Убедимся, что последняя часть заканчивается знаком препинания
    if not re.search(r'[.!?]$', parts[-1]):
        parts[-1] += '.'  # Добавляем точку, если её нет

    return parts[:5]


def build_table(text):
    # Разделение на предложения
    sentences = re.findall(r'([А-ЯЁ][^\.!?]*[\.!?])', text, re.DOTALL)

    table = defaultdict(list)
    for sent in sentences:
        parts = split_sentence(sent)
        for i, part in enumerate(parts):
            table[i].append(part.strip())

    # Формируем финальную таблицу
    columns = [table[i] for i in range(5)]
    return list(zip(*columns))  # Транспонируем для строкового представления


table = build_table(yandex_text)
for row in table:
    print(row)
```

```python
# 6.5
import random
import re
from collections import defaultdict


def build_markov_chain_model(text, prefix_length):
    model = defaultdict(lambda: defaultdict(int))
    for i in range(len(text) - prefix_length):
        prefix = text[i:i + prefix_length]
        next_char = text[i + prefix_length]
        model[prefix][next_char] += 1
    return model


def generate_text(model, prefix_length, length):
    possible_prefixes = [k for k in model.keys() if len(k) == prefix_length]
    if not possible_prefixes:
        return "Ошибка: нет префиксов заданной длины в модели."

    current_prefix = random.choice(possible_prefixes)
    result = current_prefix

    for _ in range(length - prefix_length):
        if current_prefix not in model:
            break
        next_char = random.choices(
            list(model[current_prefix].keys()),
            weights=list(model[current_prefix].values())
        )[0]
        result += next_char
        current_prefix = result[-prefix_length:]
    return result


def get_stats(vocab):
    pairs = defaultdict(int)
    for word, freq in vocab.items():
        symbols = word.split()
        for i in range(len(symbols) - 1):
            pairs[symbols[i], symbols[i + 1]] += freq
    return pairs


def merge_vocab(pair, v_in):
    v_out = {}
    bigram = re.escape(' '.join(pair))
    p = re.compile(r'(?<!\S)' + bigram + r'(?!\S)')
    for word in v_in:
        w_out = p.sub(''.join(pair), word)
        v_out[w_out] = v_in[word]
    return v_out


def get_vocab(data):
    vocab = defaultdict(int)
    for line in data:
        for word in line.split():
            vocab[' '.join(list(word)) + ' </w>'] += 1
    return vocab


def byte_pair_encoding(data, n):
    vocab = get_vocab(data)
    for i in range(n):
        pairs = get_stats(vocab)
        if not pairs:
            break
        best = max(pairs, key=pairs.get)
        vocab = merge_vocab(best, vocab)
    result = ' '.join([word.replace('</w>', '\n').replace(' ', '')
                      for word in vocab.keys()])
    return result


with open("pushkin.txt", "r") as file:
    text = file.read()

prefix_length = 5
gen_length = 200
bpe_iterations = 10

# Генерация без BPE
print("Без BPE:")
model_no_bpe = build_markov_chain_model(text, prefix_length)
generated_no_bpe = generate_text(model_no_bpe, prefix_length, gen_length)
print(generated_no_bpe)
print("\n")

# Генерация с BPE
print("С BPE:")
data = text.split('.')
bpe_text = byte_pair_encoding(data, bpe_iterations)
model_bpe = build_markov_chain_model(bpe_text, prefix_length)
generated_bpe = generate_text(model_bpe, prefix_length, gen_length)
print(generated_bpe)
print("\n")
```

```python
#7.1
import re

def parse_game(file_path):
    rooms = {}
    with open(file_path, 'r', encoding='utf-8') as file:
        content = file.read()
        
        content = '\n' + content
        split_pattern = r'(\n## \[.*?\]\(#\d+\)\n)'
        split_result = re.split(split_pattern, content)
        
        headers = split_result[1::2]
        blocks = split_result[2::2]
        
        for header, block in zip(headers, blocks):
            title_line = header.strip()
            label_match = re.search(r'\(#(\d+)\)', title_line)
            if not label_match:
                continue
            label = label_match.group(1)
            
            lines = block.strip().split('\n')
            description = ""
            actions = []
            for line in lines:
                line = line.strip()
                if not line:
                    continue
                if not line.startswith('*') and not description:
                    description = line
                elif line.startswith('*'):
                    action_match = re.match(r'\*\s*(.*?)\.\s*\[.*?\]\(#(\w+)\)', line)
                    if action_match:
                        action_text = action_match.group(1).strip()
                        next_label = action_match.group(2)
                        actions.append((action_text, next_label))
                    else:
                        actions.append((line, None))
            
            rooms[label] = {
                "title": title_line,
                "description": description,
                "actions": actions
            }
    
    return rooms

def play_game(rooms, start_label):
    current_label = start_label
    while True:
        room = rooms.get(current_label)
        if not room:
            print("Ошибка: комната не найдена.")
            break
        
        print(f"\n{room['title']}\n")
        print(room['description'])
        
        actions = room['actions']
        if not actions:
            print("Игра завершена.")
            break
        
        for idx, (action, _) in enumerate(actions, 1):
            print(f"{idx}. {action}")
        
        choice = input("\n> ").strip().lower()
        
        if choice in ['exit', 'quit', 'escape']:
            print("Игра завершена.")
            break
        
        if not choice.isdigit():
            print("Неверный выбор.")
            continue
        
        choice_num = int(choice)
        if choice_num < 1 or choice_num > len(actions):
            print("Неверный выбор.")
            continue
        
        _, next_label = actions[choice_num - 1]
        
        if next_label == 'end':
            print("Вы выбрались из лабиринта!")
            break
        elif next_label:
            current_label = next_label
        else:
            print("Действие не ведет в другую комнату.")

# Запуск игры
rooms = parse_game("game.md")
print(rooms)
play_game(rooms, "1")
```

```python
#7.2
import re

def parse_game(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        content = file.read()
    
    content = '\n' + content  # Добавляем \n для разделения
    split_pattern = r'(\nКомната #\d+\n)'
    split_result = re.split(split_pattern, content)
    
    headers = split_result[1::2]
    blocks = split_result[2::2]
    
    rooms = {}
    for header, block in zip(headers, blocks):
        label_match = re.search(r'Комната #(\d+)', header)
        if not label_match:
            continue
        label = label_match.group(1)
        
        lines = block.strip().split('\n')
        description_lines = []
        actions = []
        in_description = True
        
        for line in lines:
            line = line.strip()
            if not line:
                continue
            if line.startswith('Проход на'):
                in_description = False
                # Исправленное регулярное выражение
                action_match = re.match(
                    r'Проход на (.*?)\.\s*(.*?)(?:\s+\(\#(\d+)\))?\.?$',
                    line
                )
                if action_match:
                    direction = action_match.group(1).strip()
                    description_part = action_match.group(2).strip()
                    next_label = action_match.group(3)
                    
                    # Обработка выхода из лабиринта
                    if "Вы выбрались из лабиринта!" in description_part:
                        next_label = "end"
                    
                    actions.append((
                        f"Проход на {direction}. {description_part}",
                        next_label
                    ))
                else:
                    actions.append((line, None))
            elif in_description:
                description_lines.append(line)
        
        description = ' '.join(description_lines)
        rooms[label] = {
            "title": header.strip(),
            "description": description,
            "actions": actions
        }
    
    return rooms


def play_game(rooms, start_label):
    current_label = start_label
    while True:
        room = rooms.get(current_label)
        if not room:
            print("Ошибка: комната не найдена.")
            break
        
        # Показываем текущую комнату
        print(f"\n{room['title']}\n")
        print(room['description'])
        
        actions = room['actions']
        if not actions:
            print("Игра завершена: нет доступных действий.")
            break
        
        # Выводим действия с номерами
        for idx, (action, _) in enumerate(actions, 1):
            print(f"{idx}. {action}")
        
        # Получаем выбор игрока
        choice = input("\n> ").strip().lower()
        
        # Проверка команд выхода
        if choice in ['exit', 'quit', 'escape']:
            print("Игра завершена.")
            break
        
        # Проверка корректности ввода
        if not choice.isdigit():
            print("Пожалуйста, введите номер действия.")
            continue
        
        choice_num = int(choice)
        if choice_num < 1 or choice_num > len(actions):
            print("Неверный выбор.")
            continue
        
        # Выполняем действие
        _, next_label = actions[choice_num - 1]
        
        if next_label == 'end':
            print("Вы выбрались из лабиринта! Поздравляем!")
            break
        elif next_label:
            current_label = next_label
        else:
            print("Этот путь никуда не ведёт.")

rooms = parse_game("game.txt")
play_game(rooms, "1")
```

```python
#7.5
import re
from collections import defaultdict, deque

def parse_game(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        content = file.read()
    
    content = '\n' + content
    split_pattern = r'(\nКомната #\d+\n)'
    split_result = re.split(split_pattern, content)
    
    headers = split_result[1::2]
    blocks = split_result[2::2]
    
    rooms = {}
    for header, block in zip(headers, blocks):
        label_match = re.search(r'Комната #(\d+)', header)
        if not label_match:
            continue
        label = label_match.group(1)
        
        lines = block.strip().split('\n')
        description_lines = []
        actions = []
        in_description = True
        
        for line in lines:
            line = line.strip()
            if not line:
                continue
            if line.startswith('Проход на'):
                in_description = False
                action_match = re.match(
                    r'Проход на (.*?)\.\s*(.*?)(?:\s+\(\#(\d+)\))?\.?$',
                    line
                )
                if action_match:
                    direction = action_match.group(1).strip()
                    description_part = action_match.group(2).strip()
                    next_label = action_match.group(3)
                    
                    if "Вы выбрались из лабиринта!" in description_part:
                        next_label = "end"
                    
                    actions.append((f"Проход на {direction}. {description_part}", next_label))
                else:
                    actions.append((line, None))
            elif in_description:
                description_lines.append(line)
        
        description = ' '.join(description_lines)
        rooms[label] = {
            "title": header.strip(),
            "description": description,
            "actions": actions
        }
    
    return rooms

def find_dead_ends(rooms):
    reverse_graph = defaultdict(list)
    for room_label in rooms:
        current_room = rooms[room_label]
        for action in current_room['actions']:
            desc, next_label = action
            if next_label is None:
                continue
            if next_label == 'end':
                reverse_graph['end'].append(room_label)
            else:
                reverse_graph[next_label].append(room_label)
    
    visited = set()
    queue = deque()
    queue.append('end')
    visited.add('end')
    
    while queue:
        current_node = queue.popleft()
        for neighbor in reverse_graph[current_node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    
    dead_ends = []
    for label in rooms:
        if label not in visited:
            dead_ends.append(label)
    return dead_ends


rooms = parse_game("game.txt")
dead_ends = find_dead_ends(rooms)
print("Тупики:", dead_ends)
```
