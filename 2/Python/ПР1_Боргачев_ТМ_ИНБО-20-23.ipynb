````python
#3.4
import random

def naive_mul(x, y):
    negative = -1
    negative = negative + 2 * ((x<0 and y<0) or (x>0 and y>0))
    r = 0 
    x =abs(x)
    y = abs(y)
    while x>0 :
        if x % 2 != 0: 
            r += y  
        x //= 2  
        y *= 2
    
    return r * negative

def test_naive_mul():
    for _ in range(1000):
        x = random.randint(-1000, 1000) 
        y = random.randint(-1000, 1000)
        
        assert naive_mul(x, y) == x * y, f"Ошибка при x={x}, y={y}"
    
    print("Все тесты пройдены успешно!")


test_naive_mul()
````

````python
#3.5
def fast_mul(x,y):
    negative = -1
    negative = negative + 2 * ((x<0 and y<0) or (x>0 and y>0))
    r = 0 
    x = bin(abs(x))[2::]
    y = bin(abs(y))[2::]
    for i in range(len(y)):
        if y[-(i + 1)] == '1':
            r += int(x,base=2) << i
    return r * negative

def test_fast_mul():
    for _ in range(1000):
        x = random.randint(-1000, 1000) 
        y = random.randint(-1000, 1000)
        
        assert fast_mul(x, y) == x * y, f"Ошибка при x={x}, y={y}"
    
    print("Все тесты пройдены успешно!")

        
test_fast_mul()
````

````python
#3.6
def fast_pow(x,y):
    if(y==0):
        return 1
    negative_pow = y<0
    r = 1
    y = abs(y)
    while y > 0:
        if y & 1:
            r *= x
        x *= x
        y >>= 1
    if (negative_pow):
        r = 1/r
    return r
def test_fast_pow():
    for _ in range(1000):
        x = random.randint(-1000, 1000) 
        y = random.randint(-1000, 1000)
        
        assert fast_pow(x, y) == x ** y, f"Ошибка при x={x}, y={y}"
    
    print("Все тесты пройдены успешно!")

        
test_fast_pow()
````

````python
#3.7
def mul_bits(x, y, bits):
    x &= (2 ** bits - 1)
    y &= (2 ** bits - 1)
    return x * y
def mul16(x, y):
    negative = -1
    negative = negative + 2 * ((x<0 and y<0) or (x>0 and y>0))
    x = abs(x)
    y = abs(y)
    x_1 = (x >> 8) & 0xFF  
    x_2 = x & 0xFF         
    y_1 = (y >> 8) & 0xFF  
    y_2 = y & 0xFF         
    
    term1 = mul_bits(x_1, y_1, 8)  
    term2 = mul_bits(x_1, y_2, 8) 
    term3 = mul_bits(x_2, y_1, 8)  
    term4 = mul_bits(x_2, y_2, 8)  
    
    result = (term1 << 16) + ((term2 + term3) << 8) + term4
    
    return result * negative
def test_mul16():
    for _ in range(1000):
        x = random.randint(-1000, 1000) 
        y = random.randint(-1000, 1000)
        
        assert mul16(x, y) == x * y, f"Ошибка при x={x}, y={y}"
    
    print("Все тесты пройдены успешно!")

        
test_mul16()
````

````python
#3.8
def mul16k(x, y):
    negative = -1
    negative = negative + 2 * ((x<0 and y<0) or (x>0 and y>0))
    x = abs(x)
    y = abs(y)
    x_1 = (x >> 8) & 0xFF  
    x_2 = x & 0xFF         
    y_1 = (y >> 8) & 0xFF  
    y_2 = y & 0xFF         
    

    ac = x_1 * y_1
    bd = x_2 * y_2
    ad_bc = (x_1 + x_2) * (y_1 + y_2) - ac - bd  # (a+b)(c+d) - ac - bd
    result = (ac << 16) + (ad_bc << 8) + bd
    
    return result * negative

def test_mul16k():
    for _ in range(1000):
        x = random.randint(-1000, 1000) 
        y = random.randint(-1000, 1000)
        
        assert mul16k(x, y) == x * y, f"Ошибка при x={x}, y={y}"
    
    print("Все тесты пройдены успешно!")

        
test_mul16k()
````

````python
#3.9
def fast_mul_gen(y):

    print(f"def f(x):")
    print(f"    r = 0")
    
    binary_y = bin(abs(y))[2:]
    binary_length = len(binary_y)
    
    temp_var = "x"
    
    for i in range(binary_length):
        if binary_y[-(i + 1)] == '1':  #Если бит 1, добавляем к результату
            print(f"    r += {temp_var}  ")
        if i < binary_length - 1:
            new_temp_var = f"x{i+1}"
            print(f"    {new_temp_var} = {temp_var} + {temp_var}  # x * 2^{i+1}")
            temp_var = new_temp_var
    
    if y < 0:
        print(f"    r = -r")
    
    print(f"    return r")


print("Генерация программы для умножения на 18:")
fast_mul_gen(18)

print("\nГенерация программы для умножения на 17:")
fast_mul_gen(17)
````

````python
#6.3
import random

# Списки имен и отчеств
first_names = ["Данил", "Роман", "Лев", "Ильдар", "Николай", "Данила", "Семен", "Самир", "Тамерлан", "Святослав", "Герман", "Назар", "Степан", "Леонид"]
middle_names = ["А.", "Д.", "Ц.", "Ш.", "Р.", "И.", "Л.", "З.", "К.", "Е.", "Н.", "О."]

# Генерация фамилий
def generate_surname():
    consonants = "бвгджзклмнпрстфхцчшщ"
    vowels = "аеёиоуыэюя"
    length = random.randint(4, 8)
    surname = "".join(random.choice(consonants if i % 2 == 0 else vowels) for i in range(length))
    return surname.capitalize() + random.choice(["ов", "ев", "ин", "ян", "ский"])

# Генерация ФИО
def generate_full_name():
    first_name = random.choice(first_names)
    middle_name = random.choice(middle_names)
    surname = generate_surname()
    return f"{first_name} {middle_name} {surname}"

# Пример использования
for _ in range(10):
    print(generate_full_name())
````

````python
#6.4
import random
from collections import defaultdict

def build_markov_model(text, prefix_length):
    model = defaultdict(lambda: defaultdict(int))
    for i in range(len(text) - prefix_length):
        prefix = text[i:i + prefix_length]
        next_char = text[i + prefix_length]
        model[prefix][next_char] += 1
    return model

def generate_text(model, prefix_length, length):
    current_prefix = random.choice(list(model.keys()))
    result = current_prefix
    
    for _ in range(length):
        if current_prefix not in model:
            break
        next_char = random.choices(
            list(model[current_prefix].keys()),
            weights=list(model[current_prefix].values())
        )[0]
        result += next_char
        current_prefix = result[-prefix_length:]
    
    return result

# Пример использования
text = "На дворе трава, на траве дрова"
model = build_markov_model(text, prefix_length=2)
print(generate_text(model, prefix_length=2, length=50))
````

````python
#6.5
from collections import Counter

def get_stats(vocab):
    pairs = Counter()
    for word, freq in vocab.items():
        symbols = word.split()
        for i in range(len(symbols) - 1):
            pairs[symbols[i], symbols[i + 1]] += freq
    return pairs

def merge_vocab(pair, v_in):
    v_out = {}
    bigram = " ".join(pair)
    replacement = "".join(pair)
    for word, freq in v_in.items():
        w_out = word.replace(bigram, replacement)
        v_out[w_out] = freq
    return v_out

def bpe_algorithm(text, num_merges):
    vocab = Counter(text.split())
    for _ in range(num_merges):
        pairs = get_stats(vocab)
        if not pairs:
            break
        best_pair = max(pairs, key=pairs.get)
        vocab = merge_vocab(best_pair, vocab)
    return vocab

# Пример использования
text = "low low low lowest new newer newest"
vocab = bpe_algorithm(text, num_merges=10)
print(vocab)
````

````python
#7.2
import re

def parse_game_txt(file_path):
    rooms = {}
    with open(file_path, 'r') as file:
        content = file.read()
        room_blocks = re.split(r'^---$', content, flags=re.MULTILINE)
        for block in room_blocks:
            lines = block.strip().split('\n')
            if not lines:
                continue
            title = lines[0]
            label = lines[1]
            description = lines[2]
            actions = []
            for line in lines[3:]:
                match = re.match(r"(\d+)\.\s+(.+?)\s+->\s+(\w+)", line)
                if match:
                    action_id, action_desc, next_label = match.groups()
                    actions.append((action_desc, next_label))
            rooms[label] = {"title": title, "description": description, "actions": actions}
    return rooms

def play_game_txt(rooms, start_label):
    current_label = start_label
    while True:
        room = rooms[current_label]
        print(f"\n{room['title']}\n")
        print(room['description'])
        for idx, (action, _) in enumerate(room['actions'], 1):
            print(f"{idx}. {action}")
        choice = input("> ")
        if not choice.isdigit() or int(choice) > len(room['actions']):
            print("Неверный выбор.")
            continue
        _, next_label = room['actions'][int(choice) - 1]
        current_label = next_label

# Пример использования
rooms = parse_game_txt("game.txt")
play_game_txt(rooms, "start")
````

````python
#7.3
def generate_dot_graph(rooms, output_file="game.dot"):
    dot_content = ["digraph GameWorld {", "node [shape=box];"]
    
    for label, room in rooms.items():
        dot_content.append(f'"{label}" [label="{room["title"]}"];')
        for action, next_label in room["actions"]:
            dot_content.append(f'"{label}" -> "{next_label}" [label="{action}"];')
    
    dot_content.append("}")
    
    with open(output_file, 'w') as file:
        file.write("\n".join(dot_content))

# Пример использования
generate_dot_graph(rooms)
print("Граф сохранен в game.dot")
````

````python
#7.4
def generate_html_game(rooms, output_dir="web_game"):
    import os
    os.makedirs(output_dir, exist_ok=True)
    
    # Главная страница
    index_html = """<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Text Adventure</title>
</head>
<body>
    <h1>Text Adventure Game</h1>
    <p>Welcome to the game! Start your adventure <a href="start.html">here</a>.</p>
</body>
</html>"""
    with open(os.path.join(output_dir, "index.html"), 'w') as file:
        file.write(index_html)
    
    # Страницы комнат
    for label, room in rooms.items():
        room_html = f"""<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>{room["title"]}</title>
</head>
<body>
    <h1>{room["title"]}</h1>
    <p>{room["description"]}</p>
    <ul>
"""
        for action, next_label in room["actions"]:
            room_html += f'<li><a href="{next_label}.html">{action}</a></li>\n'
        room_html += "</ul>\n</body>\n</html>"
        
        with open(os.path.join(output_dir, f"{label}.html"), 'w') as file:
            file.write(room_html)

# Пример использования
generate_html_game(rooms)
print("HTML-проект сохранен в папке web_game")
````

````python
#7.5
def find_dead_ends(rooms):
    visited = set()
    stack = ["start"]
    
    while stack:
        current = stack.pop()
        if current in visited:
            continue
        visited.add(current)
        for _, next_label in rooms[current]["actions"]:
            stack.append(next_label)
    
    dead_ends = [label for label in rooms if label not in visited]
    return dead_ends

# Пример использования
dead_ends = find_dead_ends(rooms)
if dead_ends:
    print("Обнаружены тупики:", dead_ends)
else:
    print("Тупиков не обнаружено.")
````

````python
#8.1
import argparse
import os

def list_directory(path, show_hidden=False, long_format=False):
    entries = os.listdir(path)
    if not show_hidden:
        entries = [e for e in entries if not e.startswith('.')]
    if long_format:
        entries = [(e, os.stat(os.path.join(path, e))) for e in entries]
        entries = [f"{e[0]} ({e[1].st_size} bytes)" for e in entries]
    return entries

parser = argparse.ArgumentParser(description="List directory contents.")
parser.add_argument("path", nargs='?', default='.', help="Directory path")
parser.add_argument("-a", "--all", action="store_true", help="Show hidden files")
parser.add_argument("-l", "--long", action="store_true", help="Long listing format")

args = parser.parse_args()
for entry in list_directory(args.path, args.all, args.long):
    print(entry)
````

````python
#8.2
import os

def generate_directory_tree(path, output_file="tree.dot"):
    dot_content = ["digraph DirectoryTree {", "node [shape=box];"]
    
    def traverse_directory(current_path, parent="root"):
        for entry in os.listdir(current_path):
            full_path = os.path.join(current_path, entry)
            node_name = full_path.replace("/", "_").replace("\\", "_")
            dot_content.append(f'"{parent}" -> "{node_name}" [label="{entry}"];')
            if os.path.isdir(full_path):
                traverse_directory(full_path, node_name)
    
    traverse_directory(path)
    dot_content.append("}")
    
    with open(output_file, 'w') as file:
        file.write("\n".join(dot_content))

# Пример использования
generate_directory_tree(".")
print("Дерево каталогов сохранено в tree.dot")
````

````python
#8.3
import sys
import re

def convert_quotes(text):
    in_code_block = False
    result = []
    quote_stack = []
    
    for line in text.splitlines():
        if line.startswith("```"):
            in_code_block = not in_code_block
            result.append(line)
            continue
        
        if in_code_block:
            result.append(line)
            continue
        
        new_line = ""
        for char in line:
            if char == '"':
                if not quote_stack or quote_stack[-1] == "«":
                    new_line += "«"
                    quote_stack.append("«")
                else:
                    new_line += "»"
                    quote_stack.pop()
            else:
                new_line += char
        result.append(new_line)
    
    return "\n".join(result)

def process_markdown(input_file=None, output_file=None):
    if input_file:
        with open(input_file, 'r') as file:
            text = file.read()
    else:
        text = sys.stdin.read()
    
    converted_text = convert_quotes(text)
    
    if output_file:
        with open(output_file, 'w') as file:
            file.write(converted_text)
    else:
        print(converted_text)

# Пример использования
process_markdown("input.md", "output.md")
````
